### Step 1. Find the file food_100.csv in the folder where this file is.

###  Step 2. Open the file in Visual Studio Code and take a look.

### Step 3. Use the file and assign it to a dataframe called food

```python
import pandas as pd  # import pandas for data manipulation / importar pandas para manipulación de datos

# Read CSV from same folder and store in 'food' / Leer CSV de la misma carpeta y guardar en 'food'
food = pd.read_csv('food_100.csv', index_col=0)
food
```

### Step 4. See the first 5 entries

```python
# head() defaults to first 5 rows / head() muestra las primeras 5 filas por defecto
food.head()
```

### Step 5. What is the number of observations in the dataset?

```python
# len() returns total number of rows / len() devuelve el número total de filas
len(food)
```

### Step 6. What is the number of columns in the dataset?

```python
# shape[1] is the column count / shape[1] es el número de columnas
food.shape[1]
```

### Step 7. Print the name of all the columns.

```python
# columns attribute holds all column labels / el atributo columns contiene todas las etiquetas de columna
food.columns
```

### Step 8. What is the name of 105th column?

```python
# columns is 0-indexed so 105th column is at index 104 / columns es 0-indexado, la columna 105 está en índice 104
food.columns[104]
```

### Step 9. What is the type of the observations of the 105th column?

```python
# Select column by position and read its dtype / Seleccionar columna por posición y leer su dtype
food.iloc[:, 104].dtype
```

### Step 10. How is the dataset indexed?

```python
# index property shows the index type and range / la propiedad index muestra el tipo e intervalo del índice
food.index
```

### Step 11. What is the product name of the 19th observation?

```python
# iloc uses 0-based position; row 18 = 19th observation / iloc usa posición base 0; fila 18 = observación 19
food.iloc[18]['product_name']
```
