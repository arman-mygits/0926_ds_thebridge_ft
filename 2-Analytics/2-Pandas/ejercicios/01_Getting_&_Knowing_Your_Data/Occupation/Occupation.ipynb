### Step 12. How many different occupations are in this dataset? How many users per occupation are in the dataset?

### Step 17. What is the mean age of users?

```python
# Mean of the age column / Media de la columna age
users['age'].mean()
```

### Step 18. What is the age with least occurrence?

```python
# Count each age, sort ascending and pick the first / Contar cada edad, ordenar ascendente y tomar la primera
age_counts = users['age'].value_counts()  # frequency of each age / frecuencia de cada edad
min_count = age_counts.min()              # lowest frequency / frecuencia más baja
age_counts[age_counts == min_count]       # ages with that minimum count / edades con ese mínimo
```
